```python
import pandas as pd
import numpy as np
```

```python
# Load dataset
iris = pd.read_csv('iris.csv')

# Display first 5 rows
iris.head()
```

```python
# Dataset information
iris.info()
```

```python
# Group data by species
grouped = iris.groupby('species')

# Calculate summary statistics
summary = grouped.mean()
print(summary)
```

```python
# Summary statistics
stats = grouped.agg(['mean', 'median', 'min', 'max', 'std'])
print(stats)
```

```python
# Create list for sepal length grouped by species
setosa = iris[iris['species'] == 'Iris-setosa']['sepal_length'].tolist()
versicolor = iris[iris['species'] == 'Iris-versicolor']['sepal_length'].tolist()
virginica = iris[iris['species'] == 'Iris-virginica']['sepal_length'].tolist()

print('Setosa:', setosa)
print('Versicolor:', versicolor)
print('Virginica:', virginica)
```
