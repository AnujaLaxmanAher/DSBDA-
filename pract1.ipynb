## 7. Data Normalization
Normalization changes values to a common scale.
We use it so large values and small values become easier to compare.

```python
# We import MinMaxScaler to scale numeric values between 0 and 1.
from sklearn.preprocessing import MinMaxScaler

# These are the numeric columns we want to normalize.
cols = ["SepalLengthCm","SepalWidthCm","PetalLengthCm","PetalWidthCm"]

# We create the scaler object.
scaler = MinMaxScaler()

# fit_transform() learns the min and max values and scales the data.
# Then we store the normalized values back into the same columns.
df[cols] = scaler.fit_transform(df[cols])
df.head()
```

```python
# After normalization, min() checks the smallest value in each selected column.
# It should be close to 0.
df[cols].min()
```

```python
# max() checks the largest value in each selected column.
# It should be close to 1.
df[cols].max()
```

## 8. Convert Categorical Data (Species)
The Species column has text values.
Machines understand numbers better, so we convert these categories into numbers.

```python
# unique() shows all different category names in Species.
# We use it to see the possible flower classes.
df['Species'].unique()
```

```python
# We import LabelEncoder to convert text labels into numbers.
from sklearn.preprocessing import LabelEncoder
 
 # We create the label encoder object.
label_encoder = LabelEncoder()
 
 # We make a copy so the original df stays safe.
df_label = df.copy()
 
 # fit_transform() gives a number to each Species value.
 # This is the function that changes flower names into 0, 1, and 2.
 # Example: Iris-setosa may become 0, Iris-versicolor may become 1, and Iris-virginica may become 2.
df_label['Species'] = label_encoder.fit_transform(df_label['Species'])
```

```python
# head() shows the first 5 rows after label encoding.
# Now Species is shown as numbers instead of text.
df_label.head()
```

```python
# unique() now shows the encoded class values.
# We expect values like 0, 1, and 2.
df_label['Species'].unique()
```

## 9. One-Hot Encoding
One-hot encoding makes separate columns for each category.
It helps represent categories without keeping them in one single numeric column.

```python
# get_dummies() creates separate columns from the Species column.
 # This is the function that creates True and False values.
 # If a row belongs to Species_0, then Species_0 becomes True and the other species columns become False.
 # Because Species has 0, 1, and 2, it creates Species_0, Species_1, and Species_2.
df_encoded = pd.get_dummies(df_label, columns=['Species'])
 
 # head() shows the first 5 rows after one-hot encoding.
df_encoded.head()
```

```python
# Here we display only the one-hot encoded columns.
# True means the row belongs to that class. False means it does not.
df_encoded[['Species_0', 'Species_1', 'Species_2']].head()
```

```python
# We make one more copy to convert True/False into 1/0.
df_encoded_int = df_encoded.copy()
 
 # astype(int) is the function that changes True to 1 and False to 0.
 # This gives the same one-hot result in integer form.
df_encoded_int[['Species_0', 'Species_1', 'Species_2']] = df_encoded_int[['Species_0', 'Species_1', 'Species_2']].astype(int)
 
 # Now we display the integer one-hot encoded columns.
df_encoded_int[['Species_0', 'Species_1', 'Species_2']].head()
```
