```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
iris = pd.read_csv('iris.csv')
iris.head()
```

```python
iris.info()
```

```python
iris.hist(figsize=(10,8))
plt.suptitle('Histograms of Iris Dataset Features')
plt.show()
```

```python
iris.plot(kind='box', subplots=True, layout=(2,2), figsize=(10,8))
plt.suptitle('Boxplots of Iris Dataset Features')
plt.show()
```

```python
iris.describe()
```

```python
iris.boxplot(figsize=(10,6))
plt.show()
```
