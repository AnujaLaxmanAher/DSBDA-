```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
```

```python
data = {
    "Name" : ["A","B","C","D","E"],
    "Marks" : [90,45,500,None,49],
    "Attendance" : [89,49,77,89,None]
}
df= pd.DataFrame(data)
df
```

```python
df.describe()
```

```python
#Check Missing Values
print(df.isnull().sum())
```

```python
df.dtypes
```

```python
#Handle Missing Values
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())

df['Attendance'] = df['Attendance'].fillna(df['Attendance'].median())

print(df)
```

```python
#Handle Inconsistencies
df['Marks'] = df['Marks'].astype(int)

df['Attendance'] = df['Attendance'].astype(int)

print(df.dtypes)
```

```python
df.isnull().sum()
```

```python

df
```

```python
# Identify Outliers Using IQR Method

Q1 = df['Marks'].quantile(0.25)

Q3 = df['Marks'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR

upper_limit = Q3 + 1.5 * IQR

outliers = df[(df['Marks'] < lower_limit) | (df['Marks'] > upper_limit)]

print("Lower Limit :", lower_limit)

print("Upper Limit :", upper_limit)

print("\nOutliers")

print(outliers)
```

```python
#Apply Log Transformation
df['Marks_Log'] = np.log(df['Marks'])

print(df)
```

```python
#Apply Normalization
df['Marks_Normalized'] = (df['Marks'] - df['Marks'].min()) / (df['Marks'].max() - df['Marks'].min())

print(df)
```

```python
#Apply Scaling
scaler = MinMaxScaler()

df[['Marks_Scaled']] = scaler.fit_transform(df[['Marks']])

print(df)
```
